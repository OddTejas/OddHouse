```python
import pandas as pd
import matplotlib.pyplot as plt
print("ready to go, all libs loaded")
```

```python
url = "https://en.wikipedia.org/wiki/List_of_Indian_states_and_union_territories_by_literacy_rate"
tables = pd.read_html(url, storage_options={"User-Agent":"Mozilla/5.0"})
n= len(tables)
for i in range(n):
    print (tables[i])
df=tables[1]
print(df.columns)    
```

```python
df= tables[1]
df_clean=df.loc[:,[("State/UT", "State/UT"),("PLFS report (2024)[4]","Total")]]
df_clean.columns= ["State","LiteracyRates"]
df_clean

```

```python
print("yaayyayaya dataClean successfull")
```

```python
df_clean["State"] = df_clean["State"].str.replace(r"\[.*?\]", "", regex=True)
df_clean["State"] = df_clean["State"].str.strip()
df_clean["LiteracyRates"] = pd.to_numeric(df_clean["LiteracyRates"], errors="coerce")
df_clean = df_clean.reset_index(drop=True)
df_clean
```

```python
df_clean.to_csv("plfs_literacy_2024_clean.csv", index=False)
```
