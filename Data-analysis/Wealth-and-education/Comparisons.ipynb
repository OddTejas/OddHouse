```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
gradLink="./data/graduates_averages_2023_cleaned.csv"
wealthLink="./data/gsdp_2024_clean.csv"
graduates_tables= pd.read_csv(gradLink)
wealth_tables= pd.read_csv(wealthLink)

graduates_tables
```

```python
wealth_tables
```

```python
print(wealth_tables.head())
print(graduates_tables.head())
```

```python

graduates_tables = graduates_tables.rename(columns={"State/Ut Code": "States"})

print(graduates_tables.columns)  
print(wealth_tables.columns)
```

```python
merged= pd.merge(wealth_tables, graduates_tables, on="States")

merged
```

```python
x = merged["higher_education_average"]
y = merged["StateWealth"]
plt.scatter(x,y)
plt.figure(figsize=(4, 3), dpi=90)
plt.xlabel("% of population with higher education")
plt.ylabel("State wealth in crore rupees")
plt.title("Education and wealth of Indian states")
for i, state in enumerate(merged["States"]):
    plt.text(merged["higher_education_average"][i], merged["StateWealth"][i], state, fontsize=7)
plt.show()
```
