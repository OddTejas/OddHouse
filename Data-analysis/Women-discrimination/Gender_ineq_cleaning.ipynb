## Workforce participation of men and women 2019 - plfs
The data mainly uses the workforce participation of women between ages of 15 and 29 years

```python
workforce_tables_df = pd.read_csv("data/workforce.csv")
workforce_tables_df.head()
```

```python
workforce_tables_df = workforce_tables_df.loc[: , ['state_name','gender','value']]
workforce_tables_df.head()
```

```python
females_workforce_df= workforce_tables_df[workforce_tables_df['gender'] == 'Female']
females_workforce_df=females_workforce_df.rename(columns={'value':'female_points'})
females_workforce_df.head()
```

```python
males_workforce_df= workforce_tables_df[workforce_tables_df['gender'] == 'Male']
males_workforce_df=males_workforce_df.rename(columns={'value':'male_points'})
males_workforce_df.head()
```

```python
workforce_tables_df= pd.merge(females_workforce_df, males_workforce_df, on='state_name')
workforce_tables_df=workforce_tables_df.loc[:,['state_name','female_points','male_points']]
workforce_tables_df.head()
```

```python
workforce_tables_df.to_csv('workforce_difference_women_and_men.csv',index=False)
```
