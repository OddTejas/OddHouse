# visualising the data 
## Bar graph of states with most inequality

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
index_df = pd.read_csv("oddhouse_index_ordered.csv")
index_df.head()
```

```python
plt.figure(figsize=(15,10))
sns.set_context("talk")
sns.set_theme()
sns.barplot(data=index_df, x='oddhouse_index', y= 'state_name', legend= False,hue = 'state_name', palette = 'mako')
plt.title("OddHouse gender inequality index (1.0 = most inequal)", fontsize= 15, pad= 20)
plt.xlabel("Composite score", fontsize=20,)
plt.ylabel("Indian States", fontsize=20)
sns.despine()
plt.show()
```
