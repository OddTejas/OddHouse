# Using the data to create a composite index
## Creating a ratio to find relative difference between men and women
### PLFS data
Since the data has all india and Jammu and kashmir which is an UT and is missing from plfs data we should remove it to keep the normalisation accurate

```python
import pandas as pd
workforce_df = pd.read_csv("cleaned_data/workforce_difference_women_and_men.csv")
workforce_df["state_name"] = workforce_df["state_name"].replace({"Himachal Pradesh":"Himachal"})
workforce_df["state_name"] = workforce_df["state_name"].replace({"Maharastra":"Maharashtra"})
workforce_df = workforce_df.drop(index=[0,9])
workforce_df = workforce_df.reset_index(drop=True)
workforce_df
```

To find how much more women work compared to men we shall find a ratio by women/men and then normalise the data to a scale of 0 to 1 to facilitate comparison under non apples to apples comparison
since we are finding the highest to lowest inequal states, states with high ratio near 1 have are more equal thus the normalisation index should represent them as 0 meaning least inequal state.

```python
workforce_df["ratio"] = workforce_df["female_points"] / workforce_df["male_points"]
workforce_df["normal"] =1 -( (workforce_df["ratio"] - workforce_df["ratio"].min()) / (workforce_df["ratio"].max() - workforce_df["ratio"].min()))
workforce_df
```

```python
workforce_df= workforce_df.rename(columns={"normal":"workforce_inequality"})
workforce_df.head()
```

### TUS data

```python
tus_df = pd.read_csv("cleaned_data/TUS_domestic_service.csv")
tus_df = tus_df.drop(index=[1,6,15,16,17,18,22,25])
tus_df["state_name"] = tus_df["state_name"].replace({"Madya Pradesh":"Madhya Pradesh"})
tus_df["state_name"] = tus_df["state_name"].replace({"Maharastra":"Maharashtra"})
tus_df["state_name"] = tus_df["state_name"].replace({"Gujrat":"Gujarat"})
tus_df = tus_df.reset_index(drop=True)
tus_df
```

Now we have to translate the diffrence by taking a ratio and normalising again

```python
tus_df["ratio"] = tus_df["female"] / tus_df["male"]
tus_df.head()
```

the number closer to 1 is the state with most equality this we must assign 1 to least inequal state or the states with the higest ratio number must be 1

```python
tus_df["normal"] =(tus_df["ratio"] - tus_df["ratio"].min()) / (tus_df["ratio"].max() - tus_df["ratio"].min())
tus_df
```

```python
tus_df= tus_df.rename(columns={"normal":"unpaid_labour"})
tus_df.head()
```

### political represnetation of women in states

```python
pol_df = pd.read_csv("cleaned_data/prs_political_participation.csv")
pol_df = pol_df.rename(columns={'States':'state_name'})
pol_df["state_name"] = pol_df["state_name"].replace({"Madya Pradesh":"Madhya Pradesh"})
pol_df["state_name"] = pol_df["state_name"].replace({"Maharastra":"Maharashtra"})
pol_df["state_name"] = pol_df["state_name"].replace({"Gujrat":"Gujarat"})
pol_df = pol_df.drop(index=[1,6,15,16,17,18,22,25])
pol_df = pol_df.reset_index(drop=True)
pol_df
```

Now to be consistant we need to use ratio of women%/men% but since there is no men% data we ought to subtract 1- women % to get the required value

```python
pol_df["Men_pct"] = 100 - pol_df["Women_pct"]
pol_df.head()
```

```python
pol_df["ratio"] = pol_df["Women_pct"] / pol_df["Men_pct"]
pol_df.head()
```

states with equal represtation must have 50-50 share of power and thus equal states should have ratio of 1 and unequal states have 0. Any states with ratio more than 1 mean that state has more women participation but this cannot be seen in this data

```python
pol_df["normal"] = (1- (pol_df["ratio"] - pol_df["ratio"].min()) / (pol_df["ratio"].max() - pol_df["ratio"].min()))
pol_df
```

```python
pol_df= pol_df.rename(columns={"normal":"political_inequality"})
pol_df.head()
```

```python
index_df = pd.merge(pol_df, tus_df, on='state_name', how='inner')
print(index_df.shape)
index_df = pd.merge(index_df, workforce_df, on='state_name', how='inner')
print(index_df.shape)
index_df.head()
```

```python
index_df= index_df.loc[:,['state_name','political_inequality','unpaid_labour','workforce_inequality']]
index_df.head()
```

```python
index_df["oddhouse_index"] = (index_df["political_inequality"]
                            + index_df["unpaid_labour"]
                            +index_df["workforce_inequality"])/3
oddhouse_index = index_df
oddhouse_index.to_csv("oddhouse_index.csv", index = False)
oddhouse_index = oddhouse_index.sort_values(by='oddhouse_index', ascending=False)
oddhouse_index.to_csv("oddhouse_index_ordered.csv", index = False)
oddhouse_index.head(10)
```
